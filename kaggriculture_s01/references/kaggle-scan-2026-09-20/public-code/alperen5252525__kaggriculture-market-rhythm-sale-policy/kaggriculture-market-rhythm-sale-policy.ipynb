# Kaggriculture: Market Rhythm — A Measured Sale Policy

**A sale has a quantity, a price—and a place in the queue.** This notebook
keeps the working farm core and compares when and how its ready products reach
the market. One selected agent, complete source, matched comparisons and
declared assumptions are packaged below.

The frozen selection finished **132 wins / 12 losses /
0 ties** in **144 responding local confirmation games**.
The live leaderboard rating of this new artifact is measured only after submission;
the tables here are game results and final cash margins, not predicted rating.

## The selected policy

The selected version retains the full credited V48 core, advances eligible
planned cash sales by up to **24 turns**, and applies bounded sale-order search.
Three plausible rival queues are evaluated: same order, nominal-price priority,
and reversed order. Rival stock is a bounded scenario informed by public ripe
yield and our stock, not a recovered hidden observation.

The objective is `mean(own_revenue - 0.0 * rival_revenue) - 0.25 * scenario_spread`.
Two adjacent-exchange passes change only neighboring cash-sale slots; purchases
and funding slots are not permuted by this new search. The inherited farming,
storage, herd and opening mechanisms remain in place. This is a finite scenario
heuristic, not a full-game equilibrium or a newly invented production engine.

The opening wheat-pressure quantity is also explicitly set to10 instead of30; the inherited next-turn resale is retained. This is the declared joint comparison, not an undisclosed extra feature.

Only cash-product sales already present in the future plan are advanced, with
physically projected stock checks, pickup exclusions and inherited funding
protection. The dawn guard turn is excluded and the scan stops at turn718.
No future shop unlocks, private competitor code, seed table or opponent names
enter the decision. Source SHA256:`f3be6392671608e9a575145c6caba4ab2f76d44b7ed64cdc086c8bcf845b541a`.

## Search first, freeze before confirmation

The first round screened14 configurations in280 games and retained R4.
The expanded round screened22 configurations, including live R3 and yesterday's R4,
on two development initial seeds, five responding policies and both
seats:440 games. The declared selection rule required zero wrapper errors and
preferred worst-rival match points, then overall points, then cash margin.
The initial opening10 selection then underwent432 confirmation games. Those
results motivated a joint refinement: three queue/opening combinations and
opening10 on three new development seeds,120 games. That earlier confirmation
became selection evidence, not a final untouched result.
The selected source was frozen before twelve unused initial seeds:576 games
across the distinct compared configurations, six policies and both seats.

## Matched confirmation: show gains and regressions


```python
import matplotlib.pyplot as plt
families=evidence['confirmation']['families'];selected=evidence['selected'];comparison=evidence['comparison'];names=list(families[selected])
deltas=[families[selected][o]['mean_margin']-families[comparison][o]['mean_margin'] for o in names]
fig,ax=plt.subplots(figsize=(9,3.5))
ax.barh(names,deltas,color=['#138a72' if v>=0 else '#c55757' for v in deltas]);ax.axvline(0,color='#334155',linewidth=1)
ax.set_title('Responding opponents: improvements and regressions')
ax.set_xlabel('Mean final-cash margin change versus '+comparison)
ax.spines[['top','right']].set_visible(False);fig.tight_layout();fig.savefig('comparison.png',dpi=160)
plt.show()
```

## Opening tradeoff

```python
screen=evidence['screen']['summary']
pressure=[0,10,20,30,40];names=['opening0','opening10','opening20','r4','opening40']
points=[screen[n]['points']*100 for n in names];worst=[screen[n]['worst_margin'] for n in names]
fig,axes=plt.subplots(1,2,figsize=(9,3.5))
axes[0].bar([str(q) for q in pressure],points,color='#138a72');axes[0].set_ylim(0,105)
axes[0].set_ylabel('Development match points (%)');axes[0].set_xlabel('Opening pressure quantity')
axes[1].bar([str(q) for q in pressure],worst,color=['#138a72' if v>=0 else '#c55757' for v in worst])
axes[1].set_yscale('symlog',linthresh=100);axes[1].set_ylabel('Worst final-cash margin (symlog)')
axes[1].set_xlabel('Opening pressure quantity')
for ax in axes:ax.spines[['top','right']].set_visible(False)
fig.suptitle('Opening pressure: more is not always better — development panel only')
fig.tight_layout();fig.savefig('opening_tradeoff.png',dpi=160);plt.show()
```
